```python
num = int(input("Enter a number: "))

temp = num

sum_digit = 0

reverse = 0

while temp != 0:
    digit = temp % 10
    sum_digits += digit
    reverse = reverse * 10 + digit
    temp //= 10

print("sum of digit =", sum_digits)
print("Reverse of number =", reverse)
```
